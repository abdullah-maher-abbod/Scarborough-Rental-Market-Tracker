```python
def clean_zone_table(raw, metric, roles):
    # 1. Column names
    bedrooms = raw.iloc[4].ffill()
    dates = raw.iloc[5].ffill()
    names = ["zone"]
    for col in range(1, raw.shape[1]):
        position = (col - 1) % len(roles)
        role = roles[position]
        bed = bedrooms[col]
        date = dates[col].split()[-1]
        if role == "value":
            names.append(f"{bed}_{date}")
        elif role == "q":
            names.append(f"{bed}_{date}_q")
        else:
            names.append(f"{bed}_change")
    names = [n.lower().replace("+", "plus").replace(" ", "_").replace("-", "") for n in names]

    # 2. Data rows only, cut after the last real row
    data = raw.iloc[6:].copy()
    data.columns = names
    last = data[data["zone"].str.strip() == "Toronto CMA"].index[0]
    data = data.loc[:last]

    # 3. Sort the columns into kinds
    q_cols = [c for c in data.columns if c.endswith("_q")]
    change_cols = [c for c in data.columns if c.endswith("_change")]
    value_cols = [c for c in data.columns if c not in q_cols + change_cols and c != "zone"]

    # 4. Values: numbers and flags
    values = data[["zone"] + value_cols].copy()
    
    for c in value_cols:
        values[c] = pd.to_numeric(values[c].astype(str).str.replace(",", "", regex=False), errors="coerce")
        values[c] = pd.to_numeric(values[c], errors="coerce")
    values["is_aggregate"] = ~values["zone"].str.startswith("Zone ")
    values["is_scarborough"] = values["zone"].str.contains("Scarborough")

    # 5. Reshape the values
    vac = values.melt(
        id_vars=["zone", "is_aggregate", "is_scarborough"],
        value_vars=value_cols,
        var_name="measure",
        value_name=metric,
    )
    parts = vac["measure"].str.rsplit("_", n=1, expand=True)
    vac["bedroom_type"] = parts[0]
    vac["year"] = 2000 + parts[1].str[3:].astype(int)
    vac = vac.drop(columns="measure")

    # 6. Reshape the quality letters
    if q_cols:
        q_long = data.melt(id_vars="zone", value_vars=q_cols, var_name="measure", value_name="quality")
        q_long["measure"] = q_long["measure"].str.removesuffix("_q")
        parts = q_long["measure"].str.rsplit("_", n=1, expand=True)
        q_long["bedroom_type"] = parts[0]
        q_long["year"] = 2000 + parts[1].str[3:].astype(int)
        q_long = q_long.drop(columns="measure")

    # 7. Join them and hand back the result
        return vac.merge(q_long, on=["zone", "bedroom_type", "year"], how="left")
    return vac
```
